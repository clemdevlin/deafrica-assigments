**Analytical brief: Radar vs optical crop phenology, Ejura-Sekyedumase , Ghana**

This brief compares optical (Sentinel-2 NDVI) and radar (Sentinel-1 RVI) phenology for a cropland area in Ghana over 2019 to 2020, and links the results to rainfall and land cover.

**Optical phenology.** NDVI shows two growing seasons a year. It peaked in late June to mid-July (about 0.65) and again in mid-November (0.59 to 0.67), with a dry-season minimum of 0.31 in March 2020 and a short dip to about 0.53 to 0.55 in October. This matches the area’s bimodal rainfall.

**Radar phenology.** RVI shows the same cycle: a minimum in late April (0.63), a first peak in July to August (about 0.74), a small dip in October (0.69 to 0.70), and a main peak from December to January (0.77). Low RVI corresponds to bare or newly planted fields. As the canopy builds leaves, stems and biomass, volume scattering increases and RVI rises, then falls after harvest.

**Comparison.** Both sensors identify the same stages and seasons, but radar lags NDVI by roughly one to eight weeks, because NDVI responds to leaf greenness while RVI responds to canopy structure and biomass, which keep building after greenness peaks. NDVI has the larger seasonal range (0.31 to 0.67 against 0.63 to 0.77 for RVI), while RVI is continuous because radar is unaffected by cloud during the rainy season, when optical images are scarce. The standard phenology statistics were unreliable for radar because its main peak falls on the year boundary, so peaks were read directly from the curves.

**Rainfall and land cover.** For 2021, CHIRPS rainfall preceded vegetation response by one to two months (best correlation r \= 0.67 at one month). The sharpest NDVI dip (September 2021\) coincided with the wettest month (295 mm), so it was not drought stress and may reflect cloud contamination or the gap between seasons. Land cover maps (2017 to 2023\) show cropland rising 27% to 2021, then falling to 41 km² in 2023 (39% below 2017), while built-up area grew 22%. These swings may partly reflect classification noise, so they are not treated as proof of crop failure.

**Recommendation.** Ghana’s drought early-warning systems should combine the sensors rather than rely on one. First, use radar RVI for continuous, cloud-free tracking of crop stage in the rainy season, and NDVI or EVI for greenness and stress when clear images exist. Second, use CHIRPS rainfall anomalies as the trigger: because vegetation responds one to two months after rain, a rainfall deficit gives farmers several weeks of warning before crops show stress. Third, compare each season with a multi-year baseline, and flag areas where NDVI or RVI falls well below normal while rainfall is also low. Fourth, pass alerts to farmers through extension officers and mobile messages, and validate them with field reports. Limitations are a single site, a short record, and smoothing that shifts dates by one to two weeks.
