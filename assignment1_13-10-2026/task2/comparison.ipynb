### Task 2: Optical (NDVI) vs Radar (RVI) phenology, 2019 to 2020

| Stage | Optical (NDVI) | Radar (RVI) | Radar lag |
|---|---|---|---|
| Dry-season low (bare fields, before planting) | 22 Mar 2020 (0.31) | 28 Apr 2019, 26 Apr 2020 (0.63) | ~5 weeks |
| Season 1 peak growth | 30 Jun 2019, 12 Jul 2020 (0.65) | 4 Aug 2019, 19 Jul 2020 (0.74) | ~1 to 5 weeks |
| Mid-year dip (harvest of season 1) | 20 Oct 2019 (0.55), 4 Oct 2020 (0.53) | 13 Oct 2019 (0.70), 25 Oct 2020 (0.69) | 0 to 3 weeks |
| Season 2 peak growth | 17 Nov 2019 (0.67), 15 Nov 2020 (0.59) | Dec 2019 to 19 Jan 2020 (0.77) | ~5 to 9 weeks |

*Both sensors used the same study area, time range and crop-mask setting. Radar dates were read from the smoothed RVI curve because the standard phenology table was unreliable at the year boundary.*
